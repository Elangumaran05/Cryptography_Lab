```python
import socket
from datetime import datetime

HOST = "0.0.0.0"
PORT = 8888

# Create TCP socket
server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

# Bind host and port
server.bind((HOST, PORT))

# Start listening
server.listen(5)

print("====================================")
print(" SIMPLE PYTHON HONEYPOT")
print("====================================")
print(f"Honeypot started on port {PORT}")
print("Waiting for connections...")

while True:
    # Accept connection
    client, address = server.accept()

    print("\n[+] Connection Detected")
    print("IP Address :", address[0])
    print("Port       :", address[1])
    print("Time       :", datetime.now())

    # Send message to client
    client.send(b"Welcome to Test Server\nEnter a message: ")

    # Receive message from client
    data = client.recv(1024)

    # Convert bytes into text
    message = data.decode().strip()
    print("Message    :", message)

    # Send response
    client.send(b"Message received\n")

    # Close current client
    client.close()
    print("[+] Connection Closed")
    print("------------------------------------")
```

```python
import socket

HOST = "127.0.0.1"
PORT = 8888

# Create client socket
client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

# Connect to honeypot
client.connect((HOST, PORT))

# Receive welcome message
message = client.recv(1024)
print(message.decode())

# Get input from user
user_message = input("Type Message: ")

# Send message
client.send(user_message.encode())

# Receive server response
response = client.recv(1024)
print(response.decode())

# Close connection
client.close()
```
