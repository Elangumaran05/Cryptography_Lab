```python
import socket
from datetime import datetime

HOST = "0.0.0.0"
PORT = 9999

server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
server.bind((HOST, PORT))
server.listen(5)

print(f"Honeypot started on port {PORT}")

while True:
    client, address = server.accept()

    print("\n[+] Connection Detected")
    print("IP Address :", address[0])
    print("Port       :", address[1])
    print("Time       :", datetime.now())

    client.send(b"Welcome to Server\n")

    client.close()
```

```python
import socket
# Socket = Communication endpoint between client and server.
client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
# Create an IPv4 TCP client socket. AF_INET = IPv4, SOCK_STREAM = TCP
client.connect(("127.0.0.1", 9999))

message = client.recv(1024)
# recv() means receive data. 1024 -> maximum number of bytes to receive in this call.
print(message.decode())  # .decode() = Bytes → Text

client.close()  # This closes the client's socket after communication is complete.
```
