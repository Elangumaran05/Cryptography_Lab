```python
import hashlib

# DSA Parameters (small values for demonstration)
p = 23
q = 11
g = 4

# Private and Public Keys
x = 3               # Private Key
y = pow(g, x, p)     # Public Key

# Random number
k = 7

# Input Message
msg = input("Enter Message: ")

# Hash the message
h = int(hashlib.sha256(msg.encode()).hexdigest(), 16) % q

# Signature Generation
r = pow(g, k, p) % q
k_inv = pow(k, -1, q)
s = (k_inv * (h + x * r)) % q

print("\nDigital Signature")
print("r =", r)
print("s =", s)

# Signature Verification
w = pow(s, -1, q)
u1 = (h * w) % q
u2 = (r * w) % q
v = ((pow(g, u1, p) * pow(y, u2, p)) % p) % q

print("\nVerification")
print("v =", v)

if v == r:
    print("Signature is Valid")
else:
    print("Signature is Invalid")
```
