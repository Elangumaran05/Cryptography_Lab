```python
# Demonstration of Intrusion Detection System (IDS) Using Snort
# This experiment demonstrates Snort IDS on Windows

# Step 1: Open Command Prompt as Administrator
# cd C:\\Snort

# Step 2: Verify Snort Installation
# bin\\snort.exe -V

# Step 3: Identify Network Interface
# bin\\snort.exe -W

# Step 4: Run Snort in Basic Sniffer Mode
# bin\\snort.exe -v -i 5

# Step 5: Display Packet Header and Payload
# bin\\snort.exe -vd -i 5

# Step 6: Display Packet in Hexadecimal and ASCII
# bin\\snort.exe -X -i 5

# Step 7: Packet Logging
# mkdir log
# bin\\snort.exe -dev -l log -i 5

# Step 8: Configure Snort for ICMP Detection
# cd C:\\Snort\\rules
# Create local.rules with:
# alert icmp any any -> any any (msg:"ICMP Ping Detected"; sid:1000001; rev:1;)

# Step 9: Start Snort IDS
# cd C:\\Snort
# bin\\snort.exe -A console -q -i 5 -R rules\\local.rules

# Step 10: Generate Test Traffic (in another terminal)
# ping 8.8.8.8

# Expected IDS Alert:
# [**] [1:1000001:1] ICMP Ping Detected [**]

print("Snort IDS Demonstration")
print("========================")
print("Commands used in this experiment:")
print("1. bin\\snort.exe -V                    (Verify installation)")
print("2. bin\\snort.exe -W                    (List interfaces)")
print("3. bin\\snort.exe -v -i 5               (Sniffer mode)")
print("4. bin\\snort.exe -vd -i 5              (Payload display)")
print("5. bin\\snort.exe -X -i 5               (Hex dump)")
print("6. bin\\snort.exe -dev -l log -i 5      (Packet logging)")
print("7. bin\\snort.exe -A console -q -i 5 -R rules\\local.rules  (IDS mode)")
print("8. ping 8.8.8.8                         (Test traffic)")
print()
print("IDS Alert Output:")
print("[**] [1:1000001:1] ICMP Ping Detected [**]")
```
