```python
# Tk → a GUI toolkit, inter → interface between Python and the Tk GUI toolkit
import tkinter as tk
from datetime import datetime

# Start with monitoring OFF
monitoring = False

# Start monitoring
def start_monitoring():
    global monitoring
    monitoring = True
    status_label.config(text="Status: Monitoring")  # .config() changes properties of an existing Tkinter widget
    input_box.focus_set()

# Stop monitoring
def stop_monitoring():
    global monitoring
    monitoring = False
    status_label.config(text="Status: Stopped")

# This function runs whenever a key is pressed
def monitor_key(event):
    if monitoring:
        key = event.keysym  # event contains info about the keyboard action; .keysym gives the key's name
        time = datetime.now().strftime("%H:%M:%S")
        output_box.config(state="normal")
        output_box.insert(tk.END, f"{time} Key Pressed: {key}\n")
        output_box.config(state="disabled")

# Create application window
root = tk.Tk()
root.title("Keyboard Event Monitoring")
root.geometry("600x450")

# Heading
title = tk.Label(root, text="Keyboard Event Monitoring", font=("Arial", 18, "bold"))
title.pack(pady=15)

# Input box
tk.Label(root, text="Type something here:").pack()
input_box = tk.Text(root, height=5, width=60)
input_box.pack(pady=10)

# Capture keyboard events from this box
input_box.bind("<KeyPress>", monitor_key)

# Start button
start_button = tk.Button(root, text="Start Monitoring", command=start_monitoring)
start_button.pack(pady=5)

# Stop button
stop_button = tk.Button(root, text="Stop Monitoring", command=stop_monitoring)
stop_button.pack(pady=5)

# Status
status_label = tk.Label(root, text="Status: Stopped")
status_label.pack(pady=10)

# Output area
tk.Label(root, text="Keyboard Events").pack()
output_box = tk.Text(root, height=10, width=65, state="disabled")
output_box.pack(pady=10)

# Start GUI
root.mainloop()
```
