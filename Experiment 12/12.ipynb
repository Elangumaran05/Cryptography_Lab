```python
import socket

HOST = "127.0.0.1"
PORT = 9000

# Create TCP socket
server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

# Bind server to port 9000
server.bind((HOST, PORT))

# Start listening
server.listen(1)

print("Test Server started on port 9000")
print("Waiting for MITM relay...")

# Accept connection from MITM
conn, address = server.accept()
print("Connection received from:", address)

# Receive message
data = conn.recv(1024)
message = data.decode()
print("Message received at Server:", message)

# Send reply
reply = "Hello Client, message received by Server"
conn.send(reply.encode())

# Close connections
conn.close()
server.close()
```

```python
import socket

# MITM details
MITM_HOST = "127.0.0.1"
MITM_PORT = 8000

# Server details
SERVER_HOST = "127.0.0.1"
SERVER_PORT = 9000

# Create MITM socket
mitm_server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

# Bind MITM to port 8000
mitm_server.bind((MITM_HOST, MITM_PORT))

# Start listening
mitm_server.listen(1)

print("MITM Relay started on port 8000")
print("Waiting for Client...")

# Accept client connection
client_conn, client_address = mitm_server.accept()
print("\nClient connected:", client_address)

# Receive client message
client_data = client_conn.recv(1024)
client_message = client_data.decode()
print("\nMITM observed Client message:")
print(client_message)

# Create connection to real server
server_conn = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
server_conn.connect((SERVER_HOST, SERVER_PORT))

# Forward client message
server_conn.send(client_data)
print("\nMessage forwarded to Server")

# Receive server reply
server_reply = server_conn.recv(1024)
print("\nMITM observed Server reply:")
print(server_reply.decode())

# Forward reply to client
client_conn.send(server_reply)
print("\nReply forwarded to Client")

# Close connections
server_conn.close()
client_conn.close()
mitm_server.close()
print("\nMITM Relay stopped")
```

```python
import socket

HOST = "127.0.0.1"
PORT = 8000

# Create TCP client
client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

# Connect to MITM relay
client.connect((HOST, PORT))

# Get message from user
message = input("Enter message: ")

# Send message
client.send(message.encode())

# Receive reply
reply = client.recv(1024)

# Display reply
print("Reply from Server:", reply.decode())

# Close connection
client.close()
```
