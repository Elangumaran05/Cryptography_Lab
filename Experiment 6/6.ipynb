```python
# Import subprocess to execute Linux commands
import subprocess

# Function to execute a Linux command
def run(command):
    try:
        result = subprocess.run(
            command,
            shell=True,
            text=True,
            capture_output=True
        )
        return result.stdout.strip()
    except Exception:
        return "Unable to execute command"

# Display title
print("Linux Privilege Escalation Checker")
print("-" * 40)

# Check current user
print("\n1. Current User")
print(run("whoami"))
print(run("id"))

# Check passwordless sudo permission
print("\n2. Sudo Permissions")
sudo = run("sudo -n -l 2>/dev/null")
if sudo:
    print(sudo)
else:
    print("No passwordless sudo permission found.")

# Find SUID files
print("\n3. SUID Files")
suid = run(
    "find /usr/bin /usr/sbin /bin /sbin "
    "-perm -4000 -type f 2>/dev/null"
)
if suid:
    print(suid)
else:
    print("No SUID files found.")

# Find world-writable files
print("\n4. World-Writable Files")
writable = run(
    "find /etc /usr /opt "
    "-type f -perm -0002 2>/dev/null | head -20"
)
if writable:
    print(writable)
else:
    print("No world-writable files found.")

# Check permissions of sensitive files
print("\n5. Sensitive File Permissions")
print(run("ls -l /etc/passwd /etc/shadow 2>/dev/null"))

# Find writable cron files
print("\n6. Writable Cron Files")
cron = run(
    "find /etc/cron* -type f -writable 2>/dev/null"
)
if cron:
    print(cron)
else:
    print("No writable cron files found.")

print("\nScan Completed Successfully.")
```
